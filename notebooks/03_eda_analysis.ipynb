# Mutual Fund Analytics - Day 3: Advanced Exploratory Data Analysis (EDA)

Welcome to the **Day 3 Notebook**! Today, we will conduct an advanced Exploratory Data Analysis (EDA) on our loaded SQLite database. We will analyze historical NAV prices, industry-wide monthly SIP inflows, Assets Under Management (AUM) growth, risk vs. return parameters, portfolio stock sector distributions, and investor demographics.

### Step 1: Environment Setup & Imports
We add the project root to Python's system path to access project modules and load libraries like `pandas`, `sqlite3`, `matplotlib`, and `seaborn`.

```python
import sys
from pathlib import Path

# Find project root
notebook_dir = Path.cwd()
project_root = notebook_dir.parent

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

import pandas as pd
import numpy as np
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.sans-serif'] = 'Inter'
plt.rcParams['font.family'] = 'sans-serif'
print("Libraries loaded successfully!")
```

### Step 2: Establish Database Connection
We connect to the SQLite database `mutual_fund_analytics.db` containing our cleaned tables.

```python
db_path = project_root / "data" / "db" / "mutual_fund_analytics.db"
conn = sqlite3.connect(str(db_path))
conn.execute("PRAGMA foreign_keys = ON;")
print(f"Connected to database: {db_path}")
```

### Step 3: NAV Trends Analysis
We will query the historical NAV data from `fact_nav` and compare the daily net asset value trajectories of different funds over time.

```python
query_nav = """
SELECT n.amfi_code, f.scheme_name, n.date, n.nav, f.category
FROM fact_nav n
JOIN dim_fund f ON n.amfi_code = f.amfi_code
ORDER BY n.amfi_code, n.date;
"""
df_nav = pd.read_sql_query(query_nav, conn)
df_nav['date'] = pd.to_datetime(df_nav['date'])

# Filter to a few specific schemes for clarity in plotting
schemes_to_plot = df_nav['scheme_name'].unique()[:5]
df_nav_filtered = df_nav[df_nav['scheme_name'].isin(schemes_to_plot)]

plt.figure(figsize=(12, 6))
sns.lineplot(data=df_nav_filtered, x='date', y='nav', hue='scheme_name', palette="tab10", linewidth=2.0)
plt.title("Historical NAV Trends for Selected Schemes", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Date")
plt.ylabel("Net Asset Value (NAV)")
plt.legend(title="Scheme Name", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Step 4: Industry SIP Inflows Trends
We analyze how total monthly SIP inflows correspond with the volume of active SIP accounts over time.

```python
query_sip = "SELECT * FROM fact_sip_industry ORDER BY month;"
df_sip = pd.read_sql_query(query_sip, conn)

fig, ax1 = plt.subplots(figsize=(12, 6))
sns.lineplot(data=df_sip, x='month', y='sip_inflow_crore', color='#1a73e8', linewidth=2.5, marker="o", label="SIP Inflow (Cr)", ax=ax1)
ax1.set_xlabel("Month")
ax1.set_ylabel("SIP Inflow (INR Crore)", color='#1a73e8')
ax1.tick_params(axis='y', labelcolor='#1a73e8')
plt.xticks(rotation=45)

ax2 = ax1.twinx()
sns.lineplot(data=df_sip, x='month', y='active_sip_accounts_crore', color='#34a853', linewidth=2.0, linestyle="--", marker="s", label="Active Accounts (Cr)", ax=ax2)
ax2.set_ylabel("Active SIP Accounts (Crore)", color='#34a853')
ax2.tick_params(axis='y', labelcolor='#34a853')
ax2.grid(False)

plt.title("Industry Monthly SIP Inflows vs. Active Accounts", fontsize=14, fontweight="bold", pad=15)
plt.tight_layout()
plt.show()
```

### Step 5: AUM Growth (Area Plot)
We stack AUM values of top fund houses over time to see the progression of industry assets.

```python
query_aum = "SELECT date, fund_house, aum_crore, num_schemes FROM fact_aum ORDER BY date;"
df_aum = pd.read_sql_query(query_aum, conn)

df_aum_pivot = df_aum.pivot(index='date', columns='fund_house', values='aum_crore').fillna(0)
top_5_fh = df_aum.groupby('fund_house')['aum_crore'].max().sort_values(ascending=False).index[:5]
df_aum_pivot_top = df_aum_pivot[top_5_fh]

plt.figure(figsize=(12, 6))
df_aum_pivot_top.plot(kind='area', stacked=True, color=sns.color_palette("Blues_d", 5), alpha=0.8, ax=plt.gca())
plt.title("AUM Growth of Top 5 Fund Houses", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Date")
plt.ylabel("AUM (INR Crore)")
plt.legend(title="Fund House", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Step 6: Scheme Risk vs Return Scatter Plot
We plot volatility (standard deviation) on the X-axis and 3-year return on the Y-axis to understand the risk-return profiles of different schemes.

```python
query_perf = """
SELECT scheme_name, category, return_3yr_pct, std_dev_ann_pct, sharpe_ratio, aum_crore
FROM fact_performance;
"""
df_perf = pd.read_sql_query(query_perf, conn)

plt.figure(figsize=(12, 6))
sns.scatterplot(
    data=df_perf, 
    x='std_dev_ann_pct', 
    y='return_3yr_pct', 
    hue='category', 
    size='aum_crore', 
    sizes=(40, 400), 
    palette="viridis",
    alpha=0.8
)
plt.title("Mutual Fund Risk (Volatility) vs. Return (3-Year Annualized)", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Annualized Volatility (Standard Deviation %)")
plt.ylabel("3-Year Annualized Return (%)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Step 7: Sector Concentration & Allocations
We show the cumulative stock weights held across portfolios grouped by industry sectors.

```python
query_port = """
SELECT sector, SUM(weight_pct) as total_weight
FROM fact_portfolio
GROUP BY sector
ORDER BY total_weight DESC
LIMIT 10;
"""
df_port = pd.read_sql_query(query_port, conn)

plt.figure(figsize=(10, 5))
sns.barplot(data=df_port, x='total_weight', y='sector', palette="crest")
plt.title("Top 10 Sectors by Cumulative Portfolio Weight (%)", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Cumulative Weight Across All Schemes (%)")
plt.ylabel("Sector")
plt.tight_layout()
plt.show()
```

### Step 8: Demographic Correlation Matrix
We map age groups to numeric midpoints and build a Pearson correlation heatmap to study investor transaction behavior.

```python
query_tx = """
SELECT amount_inr, annual_income_lakh, age_group
FROM fact_transactions;
"""
df_tx = pd.read_sql_query(query_tx, conn)

age_map = {
    "18-25": 21.5,
    "26-35": 30.5,
    "36-45": 40.5,
    "46-55": 50.5,
    "56+": 60.0
}
df_tx['age_midpoint'] = df_tx['age_group'].map(age_map).fillna(35.0)

numeric_df = df_tx[['amount_inr', 'annual_income_lakh', 'age_midpoint']].dropna()
plt.figure(figsize=(8, 6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".3f", linewidths=0.5)
plt.title("Correlation Matrix: Investor Profile & Transactions", fontsize=14, fontweight="bold", pad=15)
plt.tight_layout()
plt.show()
```

### Step 9: Category Net Inflow Comparisons
Finally, we compare total net inflows by category to see where capital flows.

```python
query_cat = """
SELECT category, SUM(net_inflow_crore) as total_net_inflow
FROM fact_category_inflows
GROUP BY category
ORDER BY total_net_inflow DESC;
"""
df_cat = pd.read_sql_query(query_cat, conn)

plt.figure(figsize=(10, 5))
sns.barplot(data=df_cat, x='total_net_inflow', y='category', palette="magma")
plt.title("Total Net Inflows by Fund Category (INR Crores)", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Net Inflow (INR Crore)")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```
